- `xyxy`: [x_min, y_min, x_max, y_max] – absolute pixel coordinates.
- `xyxyn`: same as `xyxy` but normalized to the image dimensions.
- `xywh`: [x_center, y_center, width, height] – absolute format.
- `xywhn`: normalized `xywh`.

Each detection row in `data` has:

[x1, y1, x2, y2, confidence, class_id]
